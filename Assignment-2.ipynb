###NumPy (Array Operations)

```python
#1
import numpy as np
import pandas as pd
from random import randint
# Create a 5×5 array with random integers from 1 to 50
arr = np.random.randint(1,51,size =(5,5))
print(arr)

## Find the row with the maximum sum
row_sum = arr.sum(axis=1)
max_row_index =  np.argmax(row_sum)

print('\nRow Sums: ')
print(row_sum)

print("\nRow with the maximum sum:")
print(arr[max_row_index])

print("Index of the row:", max_row_index)

# Calculate the mean of the array
mean_value = arr.mean()
print("\nMean of the array:", mean_value)

# Replace elements greater than the mean with 0
modified_arr = arr.copy()
modified_arr[modified_arr > mean_value] = 0

print("\nModified Array:")
print(modified_arr)
```

###Pandas (Data Handling)

```python
#3 Load the dataset Iris data o Display the first 10 rows and basic statistics. o Filter all rows where sepal_length > median and species is "versicolor". o Group the data by species and compute the mean petal_length. 

import pandas as pd
import seaborn as sns

# Step 1: Load dataset
df = sns.load_dataset("iris")

# Step 2: Display first 10 rows and basic statistics
print("First 10 rows:")
print(df.head(10))

print("\nBasic statistics:")
print(df.describe())

# Step 3: Filter rows
median_sepal_length = df['sepal_length'].median()
filtered_df = df[(df['sepal_length'] > median_sepal_length) & (df['species'] == 'versicolor')]
print("\nFiltered rows (sepal_length > median AND species = versicolor):")
print(filtered_df)

# Step 4: Group by species, mean petal_length
mean_petal_length = df.groupby('species')['petal_length'].mean()
print("\nMean petal_length by species:")
print(mean_petal_length)
```

```python
#4 Add a new column named petal_ratio = petal_length / petal_width. o Find the top 5 rows with the highest petal_ratio
df['petal_ratio'] = df['petal_length'] / df['petal_width']

# Top 5 rows with highest petal_ratio
top5 = df.nlargest(5, 'petal_ratio')
print(top5)
```

###Data Visualization


```python
#5 A histogram of sepal_length with 15 bins.
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df['sepal_length'], bins=15, color='skyblue', edgecolor='black')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Frequency')
plt.title('Histogram of Sepal Length')
plt.show()
```

```python
#5 A scatter plot of sepal_length vs petal_length, color-coded by species. 
import matplotlib.pyplot as plt

species_colors = {'setosa': 'red', 'versicolor': 'green', 'virginica': 'blue'}

plt.figure(figsize=(8, 5))
for species, color in species_colors.items():
    subset = df[df['species'] == species]
    plt.scatter(subset['sepal_length'], subset['petal_length'], 
                label=species, color=color, alpha=0.7)

plt.xlabel('Sepal Length (cm)')
plt.ylabel('Petal Length (cm)')
plt.title('Sepal Length vs Petal Length by Species')
plt.legend(title='Species')
plt.show()
```

```python
#6 Using Seaborn, plot: o A pairplot of the entire dataset. 
import seaborn as sns
import matplotlib.pyplot as plt

sns.pairplot(df, hue='species')
plt.show()
```

```python
#6 A boxplot showing petal_width distribution for each species. 
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='species', y='petal_width')
plt.xlabel('Species')
plt.ylabel('Petal Width (cm)')
plt.title('Petal Width Distribution by Species')
plt.show()
```
